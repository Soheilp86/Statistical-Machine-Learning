# Course Introduction

Welcome! This notebook is an overview of what's covered in this course.

A data science project generally moves through the same stages: get data, understand it, clean it, and then model it. The sections below walk through each stage in the order the course covers them.

## 1. Data Collection

To complete a data-based project, we need data. We will introduce a few different ways to find it:

- Data competition websites, and
- Data repositories.

## 2. Exploratory Data Analysis (EDA)

Any data science project starts with some exploratory data analysis (EDA). We will cover:

- Exploratory plotting,
- Examining basic statistics, and
- Data manipulation with `pandas` and `numpy`.

## 3. Data Cleaning

A majority of any data science project is cleaning. We will cover:

- Cleaning data files,
- Cleaning text data with `str` functionality,
- Imputing missing values,
- Creating new columns from existing columns, and
- More.

## 4. Supervised Learning Models

In supervised learning, we train a model on labeled data &mdash; examples that are paired with a known outcome &mdash; so that it can predict that outcome for new, unseen data. We will cover:

- **Regression** (predicting a continuous outcome):
    - Simple linear regression,
    - Multiple linear regression,
    - Polynomial regression,
    - Regularized regression, and
    - Time series analysis/forecasting (if time allows).
- **Classification** (predicting a category):
    - $k$-Nearest neighbors,
    - Logistic regression,
    - Decision trees,
    - Random forests, and
    - Support vector machines (if time allows).

## 5. Unsupervised Learning

In unsupervised learning, the data has no labeled outcome. Instead, we look for structure within the data itself &mdash; for example, reducing it to its most important dimensions or grouping similar observations together. We will cover:

- Principal component analysis (PCA),
- Singular value decomposition (SVD &mdash; if time allows),
- $k$-Means clustering, and
- Hierarchical clustering (if time allows).

## 6. Neural Networks (NN)

A neural network is a model built from layers of simple, connected units ("neurons") whose connection weights are learned from data. Neural networks are flexible enough to be used for both supervised learning (e.g. predicting an outcome from labeled data) and unsupervised learning (e.g. learning a compressed representation of data without labels). We will cover:

- Perceptrons,
- Multilayer perceptrons, or feed-forward networks,
- Basic convolutional networks (if time allows), and
- Recurrent neural networks (if time allows).

The main neural network approach for unsupervised learning is autoencoders, but we won't cover them in this course.
