# Jupyter Notebook

Jupyter Notebook is an interactive tool that lets you write and run code, create visualizations, and add explanations all in one place. It is widely used in data science, research, and education.

Jupyter Notebooks can be run using platforms such as Anaconda and Google Colab. In this course, we will use **Google Colab**.


## Jupyter Notebook Using Google Colab

Google Colab lets you use Jupyter Notebooks directly in your web browser, so you don’t need to install anything on your computer. You can also save your notebooks to Google Drive, making your work easy to access and share.


### Step 1: Sign In to Google Colab

1. **Google Account:**
   - You need a Gmail account to use Google Colab. If you don’t have one, you can create it [here](https://accounts.google.com/SignUp).

2. **Access Google Colab:**
   - Go to [Google Colab](https://colab.research.google.com/) and sign in with your Gmail account.

3. **Create A New Notebook:**
   - Click on the “File” menu in the top-left corner and select “New notebook.” This will open a new Jupyter notebook in the browser.

## Jupyter Notebook Using Anaconda

Anaconda is a popular platform for Python and R that makes it easy to install and manage the tools and libraries used in data science and machine learning.

With Anaconda, you can launch Jupyter Notebook directly from **Anaconda Navigator** without using the command line. It also helps you manage the packages and environments needed for your projects.


### Step 1: Install Anaconda

Download and install **Anaconda** for your operating system from the [Anaconda website](https://www.anaconda.com/products/distribution). Follow the installation instructions.

### Step 2: Open Jupyter Notebook

Open **Anaconda Navigator** and click **Launch** under **Jupyter Notebook**. Jupyter Notebook will open in your web browser.

### Step 3: Create a New Notebook

Navigate to the folder where you want to save your work. Then click **New** and select **Python 3** to create a new notebook.



## My First Jupyter Notebook (JNB)

As mentioned earlier, a Jupyter Notebook lets us combine **explanatory text** with **code** in the same document.

* **Markdown blocks** are used for text, explanations, and formatting.
* **Code blocks** are used to write and run Python code.


### Markdown Cells or Text cells

You can add explanatory text using Markdown cells. To create a Markdown cell, simply click on `+ Text` (top-right of this page). Write your text using Markdown syntax and run the cell with Shift + Enter. If you are familiar with latex, you can also write math formulas such as

$$x^2 +y^2 = 1$$

$$\sum^{n}_{i=1} x_i$$

$$∫ \sin(x) dx$$
